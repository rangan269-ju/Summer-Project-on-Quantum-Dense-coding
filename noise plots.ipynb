```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# Categories
# ==========================================

categories = [
    'Dense Coding\nCapacity',
    'Holevo\nBound',
    'Channel\nEfficiency',
    'Robustness',
    'Circuit\nDepth',
    'Max Resource\nCost'
]

N = len(categories)

angles = np.linspace(0, 2*np.pi, N, endpoint=False).tolist()
angles += angles[:1]

# ==========================================
# Original Data
# ==========================================


bell      = [2, 2, 2.00000, 0.5000, 5, 11]
ghz       = [2, 2, 1.50000, 0.3333, 8, 15]
cluster4  = [4, 4, 2.00000, 0.3333, 12, 24]
cluster5  = [4, 4, 2.50000, 0.3333, 14, 31]
cluster6  = [5, 5, 2.00000, 0.2500, 8, 18]

data = np.array([
    bell,
    ghz,
    cluster4,
    cluster5,
    cluster6
], dtype=float)

# ==========================================
# Min-Max Normalization
# ==========================================

mins = data.min(axis=0)
maxs = data.max(axis=0)

norm = (data - mins)/(maxs - mins)

labels = [
    "QDC using Bell state",
    "QDC using GHZ state",
    "QDC using Cluster-4 state",
    "QDC using Cluster-5 state",
    "QDC using Cluster-6 state"
]

colors = [
    "royalblue",
    "forestgreen",
    "darkorange",
    "crimson",
    "purple"
]

# ==========================================
# Plot
# ==========================================

fig = plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

ax.set_theta_offset(np.pi/2)
ax.set_theta_direction(-1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=11)

ax.set_ylim(0,1)

for values, label, color in zip(norm, labels, colors):

    values = values.tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=label,
        color=color
    )

    ax.fill(
        angles,
        values,
        alpha=0.12,
        color=color
    )

plt.title(
    "Comparative Performance Analysis of Quantum Dense Coding Protocols",
    fontsize=14,
    pad=25
)

plt.legend(
    loc='upper right',
    bbox_to_anchor=(1.35,1.10)
)

plt.tight_layout()
plt.show()
```
